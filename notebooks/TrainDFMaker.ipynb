```python
import pandas as pd
from pathlib import Path

df = pd.read_csv("../data/csv/homo_aff_tps_train.csv")
target_image_paths = list(Path("../data/tinyTimeMachine").rglob("*.jpg"))
```

```python
df = df[df["aff/tps/homo"] == 2] # filter only homo
df = df.sample(len(target_image_paths), random_state = 999)
for j,i in enumerate(df.index):
    df.at[i, "fname"] = str(target_image_paths[j]).split("data/")[-1]
```

```python
df.to_csv("../data/csv/homo_aff_tps_train_tinyTimeMachine.csv", index=False, sep=",")
```
